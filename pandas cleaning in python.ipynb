```python
import pandas as pd # import pandas library for data manipulation and analysis
import numpy as np # import numpy library for numerical computations
```

```python
# STEP 1: Load the dataset

df = pd.read_csv(r"C:\projects\APM\APM.csv") # read the CSV file into a pandas DataFrame
raw_df = df.copy() # create a copy of the DataFrame to preserve the original data
print("Dataset loaded successfully. Number of rows: {}, Number of columns: {}".format(df.shape[0], df.shape[1])) # print the number of rows and columns in the DataFrame
```

```python
# STEP 2: Data Cleaning and Preprocessing

# print the first 5 rows of the DataFrame to get an overview of the data
print(df.head())
```

```python
# print the shape of the DataFrame to see the number of rows and columns
print(df.shape)
```

```python
print(df.columns.tolist()) # print the column names of the DataFrame as a list
```

```python
# print the data types of each column in the DataFrame to understand the type of data being handled
print(df.dtypes)
```

```python
# print information about the DataFrame
df.info()
```

```python
df.describe() #checking the statistics of a column
```

```python
# step 3 STANDARDIZE COLUMN NAMES
df.columns = (
    df.columns
    .str.strip() # remove leading and trailing whitespace from column name
    .str.lower() # converting all text columns or string fields to lower case
    .str.replace(" ", "_") # replace space with underscore
    .str.replace("-", "_") # replace hyphen with underscore
)
df.columns.tolist() 
```

```python
# Step 5 Check for the missing values
df.isnull().sum() # for each column
```

```python
df.isnull().sum().sum() # for all column
```

```python
# Step 6 - VALIDATE DATA TYPES
df.dtypes #check for data types
```

```python
# Convert all columns to numeric where possible
for column in df.columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

print("\n--- Data Types After Conversion ---")
print(df.dtypes)

# Check whether conversion created new missing values
conversion_missing = df.isnull().sum().sum()

print("\nMissing values after type conversion:", conversion_missing)
```

```python
# STEP 7 — STATISTICS & RANGE CHECK
print("--- Descriptive Statistics ---")
print(df.describe().T)

# Check minimum and maximum values
print("--- Minimum Values ---")
print(df.min())

print("--- Maximum Values ---")
print(df.max())
```

```python
# STEP 8 — CONSTANT / LOW-VARIATION COLUMNS
unique_values = df.nunique().sort_values()

print("\n Number of Unique Values")
print(unique_values)

constant_columns = [
    column for column in df.columns
    if df[column].nunique() <= 1
]

print("\nConstant Columns")
print(constant_columns)

# We identify them first rather than automatically deleting them.
# Engineering relevance will be considered before removal.
```

```python
# STEP 9 — DATA CONSISTENCY CHECKS
print("\n--- Consistency Checks ---")

# Check engine count
engine_count = df["unit_number"].nunique()
print("Number of engines:", engine_count)

# Check cycle range
print("Minimum cycle:", df["cycle"].min())
print("Maximum cycle:", df["cycle"].max())

# Check whether cycle starts at 1 for each engine
cycle_start = df.groupby("unit_number")["cycle"].min()

print(
    "Engines not starting at cycle 1:",
    (cycle_start != 1).sum()
)

# Check for negative values
negative_values = (df < 0).sum().sum()

print("Negative numeric values:", negative_values)

# Check for duplicated engine-cycle combinations
engine_cycle_duplicates = df.duplicated(
    subset=["unit_number", "cycle"]
).sum()

print(
    "Duplicate engine-cycle combinations:",
    engine_cycle_duplicates
)

```

```python
# STEP 10 — EXPORT CLEAN DATA

output_file = "FD001_Train_Clean.csv"

df.to_csv(
    output_file,
    index=False
)

print("\n--- CLEANING COMPLETE ---")

print("Original shape:", raw_df.shape)
print("Final shape:", df.shape)

print("Output file:", output_file)

print("\nStatus: DATA READY FOR ANALYSIS")
```
